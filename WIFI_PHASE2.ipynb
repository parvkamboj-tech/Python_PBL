```python
df.hist(figsize=(10,8))
plt.show()
```

```python
sns.countplot(x="congestion", data=df)
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.show()
```

```python
sns.boxplot(x="congestion", y="latency", data=df)
plt.show()
```

```python
import seaborn as sns
sns.scatterplot(x="devices", y="latency", hue="congestion", data=df)
plt.show()
```

Key Insights:
- Higher latency increases congestion
- High bandwidth usage leads to congestion
- More devices increase network load
