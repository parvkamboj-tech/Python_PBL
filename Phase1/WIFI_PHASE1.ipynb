```python
import pandas as pd

df = pd.read_csv("UNSW_NB15_training-set.csv")
print(df.head())
```

```python
print(df.describe())
```

MISSING VALUES CHECK

```python
print(df.isnull().sum())
```

DUPLICATE REMOVES

```python
S = df.drop_duplicates(inplace=True)
print(S)
```

```python
df_new = pd.DataFrame()
```

```python
df_new["bandwidth"] = df["sbytes"] + df["dbytes"]
```

```python
df_new["latency"] = df["dur"] * 1000 #dur->duration
```

```python
df_new["packet_loss"] = (df["spkts"] - df["dpkts"]).abs()
```

```python
df_new["devices"] = df["spkts"] + df["dpkts"]
```

```python
df_new["load"] = df["sbytes"] / (df["dur"] + 1)
```

```python
df_new["congestion"] = (
    (df_new["latency"] > df_new["latency"].quantile(0.75)) |
    (df_new["bandwidth"] > df_new["bandwidth"].quantile(0.75)) |
    (df_new["load"] > df_new["load"].quantile(0.75))
).astype(int)
```
